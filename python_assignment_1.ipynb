  An interpreter is a program that converts Python source code into machine-understandable instructions one line at a time, allowing the code to run directly without compiling.

3.What are predefined keywords in python?
  Answer :
  Predefined keywords in Python are reserved words that already have a special meaning in the language.
  You cannot use them as variable names, function names, or identifiers because Python uses them to define its rules and syntax.
  Examples:if, else, while, for, True, False, class, import, return, etc.

6.Why are list mutable but tuples are immutable?
  Answer :
  Lists can change because Python allows you to edit them anytime — you can add, remove, or modify elements.
  Tuples cannot change because Python locks their values once they are created. Whatever is inside a tuple stays the same forever.

7.What is the difference between "==" and"is" operators in python?
  Answer :
  == checks if two values are the same.
  is checks if both variables point to the exact same object in memory.

  So,

  == → compares value

  is → compares identity

  Example:
  Two different lists can have the same value, so == is true, but they are stored in different places, so is is false.

9.What is type casting in python?
  Answer :
  Type casting in Python means converting one data type into another, like turning a string into an integer, or an integer into a float.
  
10.What is the difference between implicit and explicit type casting?
  Answer :
  Implicit type casting is when Python automatically converts one data type into another without you asking.
  Example: when you add an integer and a float, Python converts the integer to float automatically.

  Explicit type casting is when you manually convert a data type using functions like int(), float(), or str().
Here you decide the conversion, not Python.

12.How does the elif statement work?
   Answer :
   The elif statement is used when you want to check more than one condition.

13.What is difference between for and while loops?
   Answer :
   A for loop is used when you know exactly how many times you want to repeat something.
   It goes through each item in a sequence or repeats for a fixed range.

   A while loop is used when you don’t know the number of repetitions, and you want the loop to run as long as a condition stays true.

14.Describe a scenario where a while loop is more suitable than a for loop.
   Answer :
   A while loop is better when you don’t know how many times the loop needs to run and the repetition depends on a condition, not a count.

   Example scenario:
   Keep asking the user for a password until they enter the correct one.
   You don’t know how many attempts they will take, so a for loop is not suitable.
   A while loop keeps running as long as the password is wrong.
